Codifica en un nuevo módulo las clases necesarias para representar impedancias, resistencias, condensadores e inductancias.
Las clases deberán reutilizar las funciones desarrolladas en la Parte B cuando corresponda, importando correspondientemente

**Clase Impedancia**
Cada objeto de la clase Impedancia representará una impedancia cualquiera. Internamente almacenará un valor de tipo
complex. El constructor de la clase Impedancia permitirá crear una impedancia equivalente a partir de un identificador y un
valor complejo:
Impedancia(identificador, impedancia, tipo="Impedancia equivalente", unidad="",
valor_nominal=None, frecuencia=0)

Cada componente almacenará, como mínimo:

• Un identificador (cadena de caracteres).

• Su impedancia compleja (objeto de tipo complex).

• Su valor nominal (ohmios, faradios, henrios), cuando sea necesario. No es necesario tras asociar varios componentes.

• Su tipo: resistencia, condensador, inductancia o impedancia equivalente.

• Su unidad: ohmios, faradios, henrios, cuando sea necesario.

• La frecuencia utilizada, cuando sea necesaria.

Los valores por defecto permitirán crear una impedancia equivalente indicando únicamente su identificador y su valor
complejo, por ejemplo: Z1 = Impedancia("Z1", 5-3j)
Los atributos serán privados. Para consultar la información almacenada se utilizarán propiedades públicas de solo lectura.
Estas propiedades permitirán que otros módulos y aplicaciones (por ejemplo, la Prácitca 2) puedan consultar la información
de los objetos sin acceder directamente a sus atributos privados. Como mínimo, la clase dispondrá de las siguientes
propiedades:
• id: identificador del elemento.
• tipo: tipo de elemento ("resistencia", "condensador", "inductancia" o "Impedancia equivalente").
• valor_nominal: valor nominal del componente, cuando corresponda. Para una impedancia equivalente podrá tener
valor None.
• unidad: unidad correspondiente al valor nominal ("Ω", "F" o "H"), cuando corresponda.
• frecuencia: frecuencia utilizada para calcular la impedancia, cuando corresponda.
• real: parte real de la impedancia.
• imaginaria: parte imaginaria de la impedancia.
• modulo: módulo de la impedancia.
• angulo: ángulo de la impedancia expresado en radianes.
• valor: valor complejo de la impedancia.
Se deberán implementar las siguientes sobrecargas de operadores:
• Asociación en serie de dos impedancias mediante el operador -. Para ello se debe sobrecargar el operador “__sub__”.
Z3 = R1 - C1
• Asociación en paralelo de dos impedancias mediante el operador |. Para ello se debe sobrecargar el operador “__or__”
Z3 = Z1 | Z2
El resultado de ambos operadores será un nuevo objeto de la clase Impedancia. Se debe comprobar que el otro operando sea
una Impedancia.
La clase deberá disponer también de métodos públicos para:
• Calcular la intensidad correspondiente a una tensión.
• Calcular la tensión correspondiente a una intensidad.
• Obtener una representación legible de la impedancia.


**Clases derivadas**

Los componentes pasivos (resistencias, condensadores, inductancias) tienen el mismo comportamiento que una impedancia
genérica (se representan como un número complejo, se componen de la misma forma en serie, paralelo) pero se inicializan
de diferente forma. Lo adecuado es manejarlos en clases derivadas de la clase Impedancia, recibiendo mediante herencia
todos sus datos, operadores y propiedades.
Cada clase derivada define su propio constructor porque cada una de ellas se inicializa de una forma diferente:

**Clase Resistencia**

Representa una impedancia resistiva. La resistencia se indicará en ohmios y deberá ser mayor que cero. Su constructor recibirá:
Resistencia(identificador, resistencia)
**Clase Condensador**

Representa una impedancia capacitiva. La capacidad se indicará en faradios y deberá ser mayor que cero. La frecuencia deberá
ser también mayor que cero. Su constructor recibirá:
Condensador(identificador, capacidad, frecuencia)
**Clase Inductancia**

Representa una impedancia inductiva. La inductancia se indicará en henrios y deberá ser mayor que cero. La frecuencia deberá
ser también mayor que cero. Su constructor recibirá:
Inductancia(identificador, inductancia, frecuencia)
