**Complex numbers**


```python
z = 3 + 4j
print(z.real) # 3.0
print(z.imag) # 4.0
print(abs(z)) # 5.0

#Obtener el angulo
import cmath
angulo = cmath.phase(z)

modulo = 5
z = cmath.rect(modulo, angulo)
print(z)
```

**Formulas electricas**

Fórmulas de impedancia en función de la frecuencia ($f$)Resistencia ($R$):$$Z_R = R$$Condensador / Capacidad ($C$):$$Z_C = -j \frac{1}{2\pi f C} = \frac{1}{j 2\pi f C}$$Bobina / Inductancia ($L$):$$Z_L = j 2\pi f L$$
Ohm Law:
$$\mathbf{V} = \mathbf{I} \cdot \mathbf{Z}$$
